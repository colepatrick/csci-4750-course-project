# Question: what are you trying to predict or find, and why does it matter?
I want to predict if a video game's release year, platform, genre, or publisher predict its global sales, and if there is any correlation between any of the features and the sales per region.
My goal is to try to determine how successful a video game will be based on their characteristics and help identify which genres, publisher, or platform are most popular based on their relationships with sales.
I'm interested in finding this out because I am a big fan of video games and I could theoretically use my model (if it is good) to find out how successful a new game might be, although the new game will need to be presumed to have over 100,000 sales to be accurately guessed by the model.

# Data: where it comes from (with a link), how many rows and columns, what the target is if there is one.
The dataset comes from https://www.kaggle.com/datasets/gregorut/videogamesales. It has 16,599 video games with sales greater than 100,000 from 1980 to 2020 and 11 columns, including Rank (not to be used due to obvious correlation), Name, Platform, (Release) Year, Genre, Publisher, NA_Sales, EU_Sales, JP_Sales, Other_Sales, and Global_Sales. 
The target is Global_Sales, and looking at how much each feature contributes to it.

# Plan: the method or methods you'll start with, and why.
I plan to start with Linear Regression as a simple model to predict global sales. For the categorical data, I was a little lost on what to do for those, so I looked up on the internet and in our book and will be using one-hot encoding to properly handle the Publisher, Genre, and Platform categorical data in the regression model.
To compare against it, I will also do Random Forests, since it can similarly handle both numerical and categorical data with some modifications, and it may produce a more accurate model since it can handle non-linear relationships better, or at least act as a comparison to the Linear Regression one. Random Forests can also provide feature importance scores which can similarly help me see which characteristics matter most.

# Evaluation: how you'll split the data and which metric you'll report.
I plan to split the dataset as 80% in the training set (13279 datapoints), 20% in the testing set (3320 datapoints).
I can use k-fold cross-validation, with 5 folds (2656 approximately per fold) to estimate the test error of each of the models. The accuracy of the models will be based on the Root Mean Squared Error and Mean Absolute Error, and from there I can average the CV scores to estimate the test error, and then choose a model from there, and run it finally against the test set and report the final RMSE and MAE.
I can also create a baseline model that predicts the Global_Sales average so that I can compare it to these models to prove their effectiveness.
